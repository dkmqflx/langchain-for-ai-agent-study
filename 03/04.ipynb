# 1장. Custom Middleware — Wrap-style 응용: 멤버십 기반 모델 선택

`Context`의 `is_premium` 필드를 이용해 **로그인 사용자의 멤버십 등급에 따라 다른 모델을 제공**하는 Agent를 구현합니다.

`03.ipynb`(메시지 길이 기반)와의 차이점은, **외부에서 주입된 Context 정보**(`is_premium`)를  
`request.runtime.context`를 통해 읽어 모델을 결정한다는 것입니다.

| is_premium | 선택 모델 |
|:---|:---|
| `True` | `gpt-5` (강력한 모델) |
| `False` | `gpt-5-nano` (기본 모델, 비용 절약) |

> 실제 서비스에서는 `Context(user_id="...", is_premium=DB에서_조회한_결과)` 형태로 사용합니다.

```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from langchain.tools import tool
from langchain.agents import create_agent


# .env 파일에서 환경 변수 로드
load_dotenv()

# 모델 선언
model = init_chat_model("gpt-4o-mini")
```

```python
# 예시 코드
from dataclasses import dataclass

@dataclass
class Context:
    user_id: str
    is_premium: bool = False  # 프리미엄 여부 (기본값 False)
```

```python
from langchain.agents.middleware import wrap_model_call
# 예시 코드
@wrap_model_call
def membership_model_selector(request, handler):
    # 1. 런타임 컨텍스트에서 멤버십 정보 확인
    is_premium = request.runtime.context.is_premium

    # 2. 조건에 따른 모델 이름 결정
    if is_premium:
        model_name = "gpt-5"
    else:
        model_name = "gpt-5-nano"

    # 3. 새로운 모델 객체 생성
    new_model = init_chat_model(model_name)

    # 4. 기존 요청의 모델을 새로운 모델로 덮어쓰기 (override)
    new_request = request.override(model=new_model)

    # 5. 수정된 요청으로 실행 계속 진행
    return handler(new_request)
```
